```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
#Reaching

classes=["LRR Cube Small", 
         "LRG Cube Small", 
         "LRR Cube Large",
         "LRG Cube Large",
         "LRR Capsule Small",
         "LRG Capsule Small",
         "LRR Capsule Large",
         "LRG Capsule Large",
         "LRR Ball Small", 
         "LRG Ball Small", 
         "LRR Ball Large",
         "LRG Ball Large"]

               
accuracies=[41.3, 
           39.7,
           47.6,
           33.7,
           30.8,
           44.6,
           31.0,
           32.1,
           35.1,
           34.9,
           29.0,
           33.1]

stds=[0.8, 2.5, 2.1, 1.9, 5.2, 4.1, 2.9, 3.9, 1.0, 0.4, 3.1, 1.4]
```

```python
colors = []
for c in classes:
    if "Cube" in c:
        colors.append("orchid")   # Cube = blue
    elif "Capsule" in c:
        colors.append("gold")    # Capsule = green
    elif "Ball" in c:
        colors.append("lightseagreen")  # Ball = orange
        

x = np.arange(len(classes))
plt.figure(figsize=(12, 6))
bars = plt.bar(x, accuracies, yerr=stds, capsize=5, alpha=0.7, color=colors, edgecolor="black")

plt.axhline(y=33, color="red", linestyle="--", linewidth=1.5, label="Random Guessing Baseline")

plt.xticks(x, classes, rotation=45, ha="right", fontsize=16)
plt.ylabel("Accuracy (%)", fontsize=16)
plt.title("Color Classification Accuracies with Standard Deviations", fontsize=20)
plt.legend( fontsize=14)
plt.tight_layout()
plt.savefig("./color_classification")
plt.show()
```

```python
#reach
classes_reach_color=["LRR Cube Small", 
                   "LRR Cube Large",
                   "LRR Capsule Small",
                   "LRR Capsule Large",
                   "LRR Ball Small",
                   "LRR Ball Large"]

accuracies_color=[41.3, 
           47.6,
           30.8,
           31.0,
           35.1,
           29.0]

stds_color=[0.8, 2.1, 5.2, 2.9, 1.0, 3.1]

classes_reach_shape=["LRR Small", 
                   "LRR Large"]
accuracies_shape=[100, 
           100]
stds_shape=[0, 0]

classes_reach_size=["LRR Cube", 
                   "LRR Capsule",
                    "LRR Ball"]

accuracies_size=[100, 
           100,
           100]
stds_size=[0, 0, 0]
```

```python
from statistics import mean
colors = []
for c in classes_reach_color:
    if "Cube" in c:
        colors.append("lightcoral")   # Cube = blue
    elif "Capsule" in c:
        colors.append("firebrick")    # Capsule = green
    elif "Ball" in c:
        colors.append("orangered")  # Ball = orange
        
for s in classes_reach_shape:
    if "Small" in s:
        colors.append("lightgreen")
    if "Large" in s:
        colors.append("forestgreen")
        
for i in classes_reach_size:
    if "Cube" in i:
        colors.append("royalblue")   # Cube = blue
    elif "Capsule" in i:
        colors.append("slateblue")    # Capsule = green
    elif "Ball" in i:
        colors.append("mediumblue")

x = np.arange(len(classes_reach_color)+len(classes_reach_shape)+len(classes_reach_size))
plt.figure(figsize=(12, 6))
bars = plt.bar(x, accuracies_color+accuracies_shape+accuracies_size, capsize=5, alpha=0.7, color=colors, edgecolor="black")

plt.axhline(y=mean(accuracies_color), color="maroon", linestyle="--", linewidth=1.5, label="Mean Rel. Std. Color Classification")
plt.axhline(y=mean(accuracies_shape), color="yellowgreen", linestyle="--", linewidth=1.5, label="Mean Rel. Std. Shape Classification")
plt.axhline(y=mean(accuracies_size), color="cornflowerblue", linestyle="--", linewidth=1.5, label="Mean Rel. Std. Size Classification")

plt.xticks(x, classes_reach_color+classes_reach_shape+classes_reach_size, rotation=90, ha="right", fontsize=16)
plt.ylabel("Accuracy (%)", fontsize=16)
plt.title("Accuracies for Color, Shape and Size Classification on Reach Hidden States", fontsize=20)
plt.legend( fontsize=14)
plt.tight_layout()
plt.savefig("./reach_all_classifications_accs.png")
plt.show()
```

```python
#Attention Scores standard devs

color_labels=["Cube Small",
              "Cube Large", 
              "Ball Small",
              "Ball Large",
              "Capsule Small",
              "Capsule Large"]

rel_stds_color=[17.4, 17.3, 5, 27.8, 21.9, 11.9]

shape_labels=["Small",
              "Large"]

rel_stds_shape=[64.18, 39.40]

size_labels=["Cube",
             "Ball",
             "Capsule"]
rel_stds_size=[27.9, 48.5, 31.0]

```

```python
from statistics import mean
colors = []
for c in color_labels:
    if "Cube" in c:
        colors.append("lightcoral")   # Cube = blue
    elif "Capsule" in c:
        colors.append("firebrick")    # Capsule = green
    elif "Ball" in c:
        colors.append("orangered")  # Ball = orange
        
for s in shape_labels:
    if "Small" in s:
        colors.append("lightgreen")
    if "Large" in s:
        colors.append("forestgreen")
        
for i in size_labels:
    if "Cube" in i:
        colors.append("royalblue")   # Cube = blue
    elif "Capsule" in i:
        colors.append("slateblue")    # Capsule = green
    elif "Ball" in i:
        colors.append("mediumblue")

x = np.arange(len(color_labels)+len(shape_labels)+len(size_labels))
plt.figure(figsize=(12, 6))
bars = plt.bar(x, rel_stds_color+rel_stds_shape+rel_stds_size, capsize=5, alpha=0.7, color=colors, edgecolor="black")

plt.axhline(y=mean(rel_stds_color), color="maroon", linestyle="--", linewidth=1.5, label="Mean Rel. Std. Color Classification")
plt.axhline(y=mean(rel_stds_shape), color="yellowgreen", linestyle="--", linewidth=1.5, label="Mean Rel. Std. Shape Classification")
plt.axhline(y=mean(rel_stds_size), color="cornflowerblue", linestyle="--", linewidth=1.5, label="Mean Rel. Std. Size Classification")

plt.xticks(x, color_labels+shape_labels+size_labels, rotation=90, ha="right", fontsize=16)
plt.ylabel("Relative Standard Deviations (%)", fontsize=16)
plt.title("Relative Standard Deviations for Color Classification", fontsize=20)
plt.legend( fontsize=14)
plt.tight_layout()
plt.savefig("./relative_standard_devs.png")
plt.show()
```
